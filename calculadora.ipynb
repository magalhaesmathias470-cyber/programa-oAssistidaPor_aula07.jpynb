Aula 07 CALCULADORA

```python
import operator

def calculadora():
    # Mapeamento de operações para funções do módulo 'operator' (limpo e performático)
    operacoes = {
        "+": operator.add,
        "-": operator.sub,
        "*": operator.mul,
        "/": operator.truediv,
    }

    print("--- Calculadora Pythonic ---")
    print(f"Operações disponíveis: {', '.join(operacoes.keys())}")

    try:
        num1 = float(input("Primeiro número: "))
        op = input("Operação: ").strip()
        num2 = float(input("Segundo número: "))

        if op in operacoes:
            # Proteção contra divisão por zero
            if op == "/" and num2 == 0:
                return "Erro: Divisão por zero não permitida."

            resultado = operacoes[op](num1, num2)
            return f"Resultado: {resultado}"
        else:
            return "Operação inválida."

    except ValueError:
        return "Erro: Por favor, digite apenas números."

if __name__ == "__main__":
    print(calculadora())
```
