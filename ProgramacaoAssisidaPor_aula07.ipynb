```python
def eh_primo(n):
  if n < 2:
    return False
  for i in range(2, n):
    if n % i == 0:
      return False
  return True
```

```python
numeros_para_testar = list (range(1,21))
```

```python

numeros_para_testar = list(range(1, 21))
primos_encontrados = []
for num in numeros_para_testar:
  if eh_primo(num):
    primos_encontrados.append(num)
print(f"Lista original (1 a 20) : {numeros_para_testar}")
print(f"Números primos encontrados: {primos_encontrados}")
```
