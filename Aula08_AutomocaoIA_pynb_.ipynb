Crie um script Python simples para verificar a temperatura de um sistema.
Se for maior que 30, exiba um alerta de superaquecimento.

```python
temperatura = int(input("qual é a temperatura?")) # Exemplo de temperatura (pode ser alterado)

if temperatura > 30:
    print(f"ALERTA: Superaquecimento! A temperatura é de {temperatura}°C.")
else:
    print(f"A temperatura do sistema é de {temperatura}°C. Tudo normal.")
```
